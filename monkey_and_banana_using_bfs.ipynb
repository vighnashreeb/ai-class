```python
from collections import deque

def bfs():
    queue = deque()
    queue.append(("door", "window", False, []))

    while queue:
        monkey, box, banana, path = queue.popleft()

        if banana:
            print("Solution:")
            for step in path:
                print(step)
            return

        if monkey != "center":
            queue.append(("center", box, banana,
                          path + ["Monkey moves to center"]))

        if monkey == "center" and box != "center":
            queue.append(("center", "center", banana,
                          path + ["Monkey pushes box to center"]))

        if monkey == "center" and box == "center":
            queue.append(("center", "center", True,
                          path + ["Monkey climbs the box"]))

bfs()
```
